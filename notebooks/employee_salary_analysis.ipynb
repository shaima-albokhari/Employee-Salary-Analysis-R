## 6. Overtime Statistics

Calculate suitable descriptive statistics for employee overtime pay.

```r
data$Overtime <- data$Payroll_Salary - data$HR_Salary
```

```r
summary(data$Overtime)
```

## 7. Group by Agency or Department

Group employees by **agency or department** and calculate:
- Number of employees
- Average salary
- Median salary
- Total overtime pay

```r
library(dplyr)
```

```r
data %>%
  group_by(Department)
```

```r
data %>%
  group_by(Department) %>%
  summarise(
    Number_of_Employees = n(),
    Average_Salary = mean(HR_Salary, na.rm = TRUE),
    Median_Salary = median(HR_Salary, na.rm = TRUE),
    Total_Overtime = sum(Overtime, na.rm = TRUE)
  )
```

## 8. Group by Job Title

Group employees by **job title** and calculate the average salary for each job title.

```r
unique(data$Occupational_Level)
```

```r
data$Job_Title <- paste(data$Department, data$Occupational_Level)
```

```r
data %>%
  group_by(Job_Title) %>%
  summarise(
    Average_Salary = mean(HR_Salary, na.rm = TRUE)
  )
```

## 9. Highest Salaries

## 10. Salary Filtering

Filter employees whose salary is above a selected salary value.

## 11. Overtime Filtering

Filter employees whose overtime pay is above a selected value.

## 12. Histogram

Create a **Histogram** showing the distribution of employee salaries.

```r
hist(data$HR_Salary,
     main = "Distribution of Employee Salaries",
     xlab = "Salary",
     ylab = "Frequency")
```

## 13. Bar Chart

Create a **Bar Chart** comparing average salaries between agencies or departments.

```r
avg_salary <- data %>%
  group_by(Department) %>%
  summarise(
    Average_Salary = mean(HR_Salary, na.rm = TRUE)
  )
```

```r
barplot(avg_salary$Average_Salary,
        names.arg = avg_salary$Department,
        main = "Average Salary by Department",
        xlab = "Department",
        ylab = "Average Salary")
```

## 14. Boxplot

Create a **Boxplot** comparing salary distributions between departments or agencies.

```r
boxplot(HR_Salary ~ Department,
        data = data,
        main = "Salary Distribution by Department",
        xlab = "Department",
        ylab = "Salary")
```

## 15. Scatter Plot

Create a **Scatter Plot** showing the relationship between:
- Base Salary
- Overtime Pay

```r
plot(data$HR_Salary, data$Overtime,
     main = "Base Salary vs Overtime Pay",
     xlab = "Base Salary",
     ylab = "Overtime Pay",
     pch = 3)
```

## 16. Correlation

Calculate the correlation between base salary and overtime pay.

```r
cor(data$HR_Salary, data$Overtime, use = "complete.obs")
```
